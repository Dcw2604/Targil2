```python
import tensorflow as tf
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, BatchNormalization, Flatten
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.regularizers import l1, l2, l1_l2
from tensorflow.keras.optimizers import Adam, RMSprop, SGD
import pandas as pd
import numpy as np
```

```python
# Load and preprocess MNIST data
(x_train, y_train), (x_test, y_test) = mnist.load_data()
```

```python
# Normalize the data
x_train = x_train.astype('float32') / 255
x_test = x_test.astype('float32') / 255
```

```python
# Convert labels to categorical
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)
```

```python
# Build the model
model = Sequential([
    # Flatten the input (28x28 images)
    Flatten(input_shape=(28, 28)),

    # First Dense layer with 32 neurons
    Dense(32, activation='relu'),
    BatchNormalization(),
    Dropout(0.3),

    # Second Dense layer with 43 neurons (corrected from 94)
    Dense(43, activation='relu'),
    BatchNormalization(),
    Dropout(0.3),

    # Third Dense layer with 69 neurons
    Dense(69, activation='relu'),
    BatchNormalization(),
    Dropout(0.3),

    # Fourth Dense layer with 22 neurons
    Dense(22, activation='relu'),
    BatchNormalization(),
    Dropout(0.3),

    # Output layer (10 classes for digits 0-9)
    Dense(10, activation='softmax')
])
```

```python
# Compile the model
model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])
```

```python
model.summary()
```

```python

# Train the model
history = model.fit(x_train, y_train,
                   batch_size=128,
                   epochs=50,
                   validation_split=0.2,
                   verbose=1)
```

```python
# Evaluate the model
test_loss, test_accuracy = model.evaluate(x_test, y_test, verbose=0)
print(f'\nTest accuracy: {test_accuracy:.4f}')
```

```python
def create_model(config, epochs=50):
    model = Sequential()
    model.add(Flatten(input_shape=(28, 28)))

    # First layer (32 neurons)
    model.add(Dense(32 * config['dense_multiplier'],
                   activation=config['activations'][0],
                   kernel_regularizer=config['regularizer']))
    if config['use_bn']:
        model.add(BatchNormalization())
    model.add(Dropout(config['dropout']))

    # Second layer (43 neurons)
    model.add(Dense(43 * config['dense_multiplier'],
                   activation=config['activations'][1],
                   kernel_regularizer=config['regularizer']))
    if config['use_bn']:
        model.add(BatchNormalization())
    model.add(Dropout(config['dropout']))

    # Third layer (69 neurons)
    model.add(Dense(69 * config['dense_multiplier'],
                   activation=config['activations'][2],
                   kernel_regularizer=config['regularizer']))
    if config['use_bn']:
        model.add(BatchNormalization())
    model.add(Dropout(config['dropout']))

    # Fourth layer (22 neurons)
    model.add(Dense(22 * config['dense_multiplier'],
                   activation=config['activations'][3],
                   kernel_regularizer=config['regularizer']))
    if config['use_bn']:
        model.add(BatchNormalization())
    model.add(Dropout(config['dropout']))

    # Output layer
    model.add(Dense(10, activation='softmax'))

    model.compile(optimizer=config['optimizer'],
                 loss='categorical_crossentropy',
                 metrics=['accuracy'])

    history = model.fit(x_train, y_train,
                       batch_size=config['batch_size'],
                       epochs=epochs,
                       validation_split=0.2,
                       verbose=1)

    test_loss, test_accuracy = model.evaluate(x_test, y_test, verbose=0)
    return test_accuracy, history.history['val_accuracy'][-1], history
```

```python
# Define regularizer options
l2_reg = l2(0.01)
l1_l2_reg = l1_l2(l1=0.01, l2=0.01)
```

```python
# 15 different experiment configurations
experiments = [
    # All ReLU
    {
        'dense_multiplier': 2,
        'activations': ['relu', 'relu', 'relu', 'relu'],
        'regularizer': l2_reg,
        'dropout': 0.3,
        'use_bn': True,
        'optimizer': Adam(learning_rate=0.001),
        'batch_size': 128
    },

    # All Sigmoid
    {
        'dense_multiplier': 2,
        'activations': ['sigmoid', 'sigmoid', 'sigmoid', 'sigmoid'],
        'regularizer': l2_reg,
        'dropout': 0.3,
        'use_bn': True,
        'optimizer': Adam(learning_rate=0.001),
        'batch_size': 128
    },

    # Alternating ReLU-Sigmoid
    {
        'dense_multiplier': 2,
        'activations': ['relu', 'sigmoid', 'relu', 'sigmoid'],
        'regularizer': l2_reg,
        'dropout': 0.3,
        'use_bn': True,
        'optimizer': Adam(learning_rate=0.001),
        'batch_size': 128
    },

    # Alternating Sigmoid-ReLU
    {
        'dense_multiplier': 2,
        'activations': ['sigmoid', 'relu', 'sigmoid', 'relu'],
        'regularizer': l2_reg,
        'dropout': 0.3,
        'use_bn': True,
        'optimizer': Adam(learning_rate=0.001),
        'batch_size': 128
    },

    # ReLU with higher capacity
    {
        'dense_multiplier': 3,
        'activations': ['relu', 'relu', 'relu', 'relu'],
        'regularizer': l2_reg,
        'dropout': 0.4,
        'use_bn': True,
        'optimizer': Adam(learning_rate=0.001),
        'batch_size': 128
    },

    # Sigmoid with higher capacity
    {
        'dense_multiplier': 3,
        'activations': ['sigmoid', 'sigmoid', 'sigmoid', 'sigmoid'],
        'regularizer': l2_reg,
        'dropout': 0.4,
        'use_bn': True,
        'optimizer': Adam(learning_rate=0.001),
        'batch_size': 128
    },

    # Mixed pattern 1
    {
        'dense_multiplier': 3,
        'activations': ['relu', 'relu', 'sigmoid', 'sigmoid'],
        'regularizer': l2_reg,
        'dropout': 0.4,
        'use_bn': True,
        'optimizer': Adam(learning_rate=0.001),
        'batch_size': 128
    },

    # Mixed pattern 2
    {
        'dense_multiplier': 3,
        'activations': ['sigmoid', 'sigmoid', 'relu', 'relu'],
        'regularizer': l2_reg,
        'dropout': 0.4,
        'use_bn': True,
        'optimizer': Adam(learning_rate=0.001),
        'batch_size': 128
    },

    # ReLU with L1L2
    {
        'dense_multiplier': 2,
        'activations': ['relu', 'relu', 'relu', 'relu'],
        'regularizer': l1_l2_reg,
        'dropout': 0.3,
        'use_bn': True,
        'optimizer': Adam(learning_rate=0.001),
        'batch_size': 128
    },

    # Sigmoid with L1L2
    {
        'dense_multiplier': 2,
        'activations': ['sigmoid', 'sigmoid', 'sigmoid', 'sigmoid'],
        'regularizer': l1_l2_reg,
        'dropout': 0.3,
        'use_bn': True,
        'optimizer': Adam(learning_rate=0.001),
        'batch_size': 128
    },

    # ReLU without BatchNorm
    {
        'dense_multiplier': 2,
        'activations': ['relu', 'relu', 'relu', 'relu'],
        'regularizer': l2_reg,
        'dropout': 0.3,
        'use_bn': False,
        'optimizer': Adam(learning_rate=0.001),
        'batch_size': 128
    },

    # Sigmoid without BatchNorm
    {
        'dense_multiplier': 2,
        'activations': ['sigmoid', 'sigmoid', 'sigmoid', 'sigmoid'],
        'regularizer': l2_reg,
        'dropout': 0.3,
        'use_bn': False,
        'optimizer': Adam(learning_rate=0.001),
        'batch_size': 128
    },

    # Mixed with RMSprop
    {
        'dense_multiplier': 2,
        'activations': ['relu', 'sigmoid', 'relu', 'sigmoid'],
        'regularizer': l2_reg,
        'dropout': 0.3,
        'use_bn': True,
        'optimizer': RMSprop(learning_rate=0.001),
        'batch_size': 128
    },

    # Mixed with SGD
    {
        'dense_multiplier': 2,
        'activations': ['relu', 'sigmoid', 'relu', 'sigmoid'],
        'regularizer': l2_reg,
        'dropout': 0.3,
        'use_bn': True,
        'optimizer': SGD(learning_rate=0.01, momentum=0.9),
        'batch_size': 128
    },

    # Maximum capacity mixed
    {
        'dense_multiplier': 4,
        'activations': ['relu', 'sigmoid', 'relu', 'sigmoid'],
        'regularizer': l1_l2_reg,
        'dropout': 0.5,
        'use_bn': True,
        'optimizer': Adam(learning_rate=0.001),
        'batch_size': 128
    }
]
```

```python

# Run experiments
results = []
for i, config in enumerate(experiments, 1):
    print(f"\nRunning experiment {i}/15...")
    print(f"Configuration:\n"
          f"Dense Multiplier: {config['dense_multiplier']}\n"
          f"Activations: {config['activations']}\n"
          f"Regularizer: {config['regularizer'].__class__.__name__ if config['regularizer'] else 'None'}\n"
          f"Dropout: {config['dropout']}\n"
          f"BatchNorm: {config['use_bn']}\n"
          f"Optimizer: {config['optimizer'].__class__.__name__}\n"
          f"Batch Size: {config['batch_size']}")

    test_acc, val_acc, history = create_model(config)

    results.append({
        'Experiment': i,
        'Dense Multiplier': config['dense_multiplier'],
        'Activations': str(config['activations']),
        'Regularizer': config['regularizer'].__class__.__name__ if config['regularizer'] else 'None',
        'Dropout': config['dropout'],
        'BatchNorm': config['use_bn'],
        'Optimizer': config['optimizer'].__class__.__name__,
        'Learning Rate': config['optimizer'].learning_rate.numpy(),
        'Batch Size': config['batch_size'],
        'Test Accuracy': test_acc,
        'Validation Accuracy': val_acc,
        'Final Training Accuracy': history.history['accuracy'][-1]
    })

```

```python
# Convert results to DataFrame and sort by test accuracy
results_df = pd.DataFrame(results)
results_df = results_df.sort_values('Test Accuracy', ascending=False)

# Print results
print("\nAll Experiment Results (sorted by Test Accuracy):")
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)
print(results_df.to_string(index=False))

# Print best configuration
best_config = results_df.iloc[0]
print("\nBest Configuration:")
for col in results_df.columns:
    print(f"{col}: {best_config[col]}")
```
